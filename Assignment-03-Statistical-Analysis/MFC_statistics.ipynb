```python
marks = [42, 55, 67, 71, 48, 83, 76, 59, 64, 91,
         52, 68, 74, 81, 57, 63, 69, 77, 85, 46,
         61, 72, 79, 88, 54, 66, 73, 82, 58, 70]
```

```python
import pystats
```

```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
# 1. Number of observations, minimum, maximum and range

n = len(marks)
minimum = min(marks)
maximum = max(marks)
data_range = maximum - minimum

print("Number of observations:", n)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", data_range)
```

```python
# 2. Mean, median and mode

mean = pystats.mean(marks)
median = np.median(marks)

from collections import Counter

frequency = Counter(marks)
max_frequency = max(frequency.values())

if max_frequency == 1:
    mode = "No mode"
else:
    mode = [x for x, f in frequency.items() if f == max_frequency]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

Mean = 67.7
Median = 68.5
Mode = No mode

The mean and median are quite close, indicating that the marks are fairly balanced around the centre. The typical student's mark is around 68 marks

```python
# 3. Q1, Q2, Q3 and IQR

Q1 = np.percentile(marks, 25)
Q2 = np.percentile(marks, 50)
Q3 = np.percentile(marks, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

The IQR is 18.5, meaning the middle 50% of marks are spread across an interval of 18.5 marks

```python
# 4. Variance, standard deviation and coefficient of variation

variance = pystats.var(marks)
standard_deviation = pystats.sd(marks)

cv = (standard_deviation / pystats.mean(marks)) * 100

print("Variance:", variance)
print("Standard deviation:", standard_deviation)
print("Coefficient of variation:", cv, "%")
```

The standard deviation is approximately 12.57 marks, indicating a moderate amount of variation around the mean. The coefficient of variation is approximately 18.57%, which indicates that the variation in marks is relatively moderate compared with the mean.

```python
# 5. Percentiles

percentiles = np.percentile(marks, [10, 25, 50, 75, 90])

print("10th percentile:", percentiles[0])
print("25th percentile:", percentiles[1])
print("50th percentile:", percentiles[2])
print("75th percentile:", percentiles[3])
print("90th percentile:", percentiles[4])
```

10th percentile = 51.6

Approximately 10% of the students scored 51.6 marks or below, while approximately 90% scored above this level.

90th percentile = 83.2

Approximately 90% of the students scored 83.2 marks or below, while approximately 10% scored above this level.

```python
# 6. Skewness

skewness = stats.skew(marks)

print("Skewness:", skewness)
```

The skewness is approximately -0.137, which is very close to zero.

Therefore, the distribution is approximately symmetric, with a slight negative (left) skew

```python
# 7. Kurtosis

kurtosis = stats.kurtosis(marks)

print("Excess kurtosis:", kurtosis)
```

The excess kurtosis is approximately -0.765, which is negative.

Therefore, the distribution is platykurtic, meaning it is relatively flatter and has lighter tails than a normal distribution.

```python
# 8. Five-number summary and outliers

print("Minimum:", minimum)
print("Q1:", Q1)
print("Median:", Q2)
print("Q3:", Q3)
print("Maximum:", maximum)

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
```

The smallest mark is 42, which is greater than 30.5, and the largest mark is 91, which is less than 104.5.
There are no potential outliers according to the 1.5 × IQR rule.

```python
# 9. Histogram and box plot

import matplotlib.pyplot as plt

plt.hist(marks, bins=7, edgecolor='black')
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Histogram of Examination Marks")
plt.show()
```

The histogram should show that the marks are mainly concentrated around 60–80, with fewer observations at the lower and upper ends.

```python
plt.boxplot(marks, vert=False)

plt.xlabel("Marks")
plt.title("Horizontal Box Plot of Examination Marks")

plt.show()
```

The box plot shows the central 50% of marks between approximately 58.25 and 76.75 and shows no potential outliers.

10. Statistical conclusion

The examination marks of the 30 students have a mean of 67.7 and a median of 68.5, indicating that the typical performance is around 68 marks. The standard deviation is approximately 12.57, while the IQR is 18.5, showing a moderate spread in student performance. The skewness of -0.137 indicates that the distribution is approximately symmetric with a slight negative skew, while the negative excess kurtosis (-0.765) suggests a relatively flat distribution. The 1.5 × IQR rule identifies no potential outliers, indicating that there are no unusually extreme marks in the dataset.
